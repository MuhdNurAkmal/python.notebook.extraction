```python
import pandas as pd
import numpy as np

ref_system = pd.read_csv("D:/Santai Coding/Pradigma - Digital Maintenance Portal/Data Migration/SAP Integration/SHAREPOINT/sharepoint_system.csv")
ref_sub_system = pd.read_csv("D:/Santai Coding/Pradigma - Digital Maintenance Portal/Data Migration/SAP Integration/SHAREPOINT/sharepoint_sub_system.csv")
ref_sub_sub_system = pd.read_csv("D:/Santai Coding/Pradigma - Digital Maintenance Portal/Data Migration/SAP Integration/SHAREPOINT/sharepoint_sub_sub_system.csv")
ref_root_cause = pd.read_csv("D:/Santai Coding/Pradigma - Digital Maintenance Portal/Data Migration/SAP Integration/SHAREPOINT/sharepoint_root_cause.csv")

ref_system_hierarchy = pd.read_csv("D:/Santai Coding/Pradigma - Digital Maintenance Portal/Data Migration/SAP Integration/SHAREPOINT/sharepoint_system_hierarchy.csv")
```

### RSD

```python
dept = "Rolling Stock Department"

system_rsd = ref_system[ref_system['department'] == dept].copy()
sub_system_rsd = ref_sub_system[ref_sub_system['department'] == dept].copy()
sub_sub_system_rsd = ref_sub_sub_system[ref_sub_sub_system['department'] == dept].copy()
root_cause_rsd = ref_root_cause[ref_root_cause['department'] == dept].copy()
system_hierarchy_rsd = ref_system_hierarchy[ref_system_hierarchy['department'] == dept].copy()
```

```python
system_map = system_rsd.set_index('system')['ID']
sub_system_map = sub_system_rsd.set_index('sub_system')['ID']
sub_sub_system_map = sub_sub_system_rsd.set_index('sub_sub_system')['ID']
root_cause_map = root_cause_rsd.set_index('root_cause')['ID']

system_hierarchy_rsd['system_id'] = system_hierarchy_rsd['system'].map(system_map)
system_hierarchy_rsd['sub_system_id'] = system_hierarchy_rsd['sub_system'].map(sub_system_map)
system_hierarchy_rsd['sub_sub_system_id'] = system_hierarchy_rsd['sub_sub_system'].map(sub_sub_system_map)
system_hierarchy_rsd['root_cause_id'] = system_hierarchy_rsd['root_cause'].map(root_cause_map)

system_hierarchy_rsd.head()
```
