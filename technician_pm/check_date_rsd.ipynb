```python
import pandas as pd

df = pd.read_excel('taskvalidation.xlsx', sheet_name='mal version')
df.head(5)
```

```python
import pandas as pd

df['start_date'] = pd.to_datetime({
    'year': df['S_year'],
    'month': df['S_month'],
    'day': df['S_day'],
    'hour': df['S_hour'],
    'minute': df['S_minutes']
})

df['end_date'] = pd.to_datetime({
    'year': df['E_year'],
    'month': df['E_month'],
    'day': df['E_day'],
    'hour': df['E_hour'],
    'minute': df['E_minutes']
})

duration_timedelta = df['end_date'] - df['start_date']
df['duration_hours'] = (duration_timedelta.dt.total_seconds() / 3600).round(2)

df_filtered = df[['workorder', 'start_date', 'end_date', 'duration_hours']]

df_filtered.to_excel('output/rsd_startend.xlsx', index=False)

df_filtered.head(5)
```
