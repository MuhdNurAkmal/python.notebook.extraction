```python
import pandas as pd
import json, ast

df_draft = pd.read_excel("extracted/draft_psd_startend_draft.xlsx", sheet_name="Sheet1")
df_draft.head()
df_draft.shape
```

```python
df_startend = pd.read_excel("extracted/snc_startend.xlsx", sheet_name="Sheet1")
df_startend.head()
df_startend.shape
```

```python
import pandas as pd

# 1. Select the required columns from both dataframes
df_startend_subset = df_startend[['workorder_no', 'actual_start', 'actual_end']]
df_draft_subset = df_draft[['workorder_no', 'inspection_date']]

# 2. Perform the left join
df_merged = pd.merge(
    df_draft_subset, 
    df_startend_subset, 
    on='workorder_no', 
    how='left'
)

# 3. Drop duplicates
df_merged = df_merged.drop_duplicates()

# 4. Combine inspection_date with actual_start and actual_end into timestamps
# We convert both to strings, concatenate with a space, and parse as datetimes.
# errors='coerce' ensures that any missing times (from the left join) safely become NaT (Not a Time / Null).

df_merged['actual_start'] = pd.to_datetime(
    df_merged['inspection_date'].astype(str) + ' ' + df_merged['actual_start'].astype(str), 
    errors='coerce'
)

df_merged['actual_end'] = pd.to_datetime(
    df_merged['inspection_date'].astype(str) + ' ' + df_merged['actual_end'].astype(str), 
    errors='coerce'
)

# Check the results and the data types
print(df_merged.dtypes)
df_merged.head()
```

```python
df_merged.to_csv("extracted/final_psd_startend.csv", index=False)
```
