```python
import pandas as pd

file_path = "D:/Santai Coding/Pradigma - Digital Maintenance Portal/Data Migration/MaintenanceOrder_20260212100905.xlsx"

df = pd.read_excel(file_path, sheet_name="Data")
df.head(3)
```

```python
pm_rsd = df[(df["oper_workcenter"] == "RSDM") & (df["mtnorder_type"] == "YPM1")].head(4)
cm_rsd = df[(df["oper_workcenter"] == "RSDM") & (df["mtnorder_type"] == "YCM1")].head(4)

rsd = pd.concat([pm_rsd, cm_rsd])
rsd
```

```python
pm_psd = df[(df["oper_workcenter"] == "WS10") & (df["mtnorder_type"] == "YPM1")].head(4)
cm_psd = df[(df["oper_workcenter"] == "WS10") & (df["mtnorder_type"] == "YCM1")].head(4)

psd = pd.concat([pm_psd, cm_psd])
psd
```

```python
pm_snc = df[(df["oper_workcenter"] == "WS20") & (df["mtnorder_type"] == "YPM1")].head(4)
cm_snc = df[(df["oper_workcenter"] == "WS20") & (df["mtnorder_type"] == "YCM1")].head(4)

snc = pd.concat([pm_snc, cm_snc])
snc
```

```python
pm_tnm = df[(df["oper_workcenter"] == "TNM1") & (df["mtnorder_type"] == "YPM1")].head(4)
cm_tnm = df[(df["oper_workcenter"] == "TNM1") & (df["mtnorder_type"] == "YCM1")].head(4)

tnm = pd.concat([pm_tnm, cm_tnm])
tnm
```

```python
df = pd.concat([rsd, psd, snc, tnm], ignore_index=True)
df
```

```python
df.to_csv(
    "NOTIFICATION_BLK_202603030000.txt",
    sep="|",
    index=False,
    header=True,
    encoding="utf-8"
)
```
